```python
features = [

    'Gender',
    'age',

    # Medical history
      'Diabetes',
      'Hypertension',
      'CKD',
      'Respiratory Disease',
      'CAD',
      'Dyslipidemia',
      'Thyroid',

    'Diabetes duration',

    # Alcohol
    'Alcohol ever',

    # Smoking
    'Smoking habbit',


    'bp_measurement-avg_systolic',
    'bmi',
    'Waist circumference',
    'Hip circumference',
    'Waist hip ratio',
    'Central Obesity'

]

target = 'non_invasive_scoring-fib_4_risk'
```

```python
X = df[features]
y = df[target]
```

```python
for col in X.columns:

    if X[col].dtype == 'object':
        X[col] = X[col].fillna(X[col].mode()[0])

    else:
        X[col] = X[col].fillna(X[col].median())
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoders = {}

categorical_cols = [

    'Gender',

    # Medical history
    'Diabetes',
    'Hypertension',
    'CKD',
    'Respiratory Disease',
    'CAD',
    'Dyslipidemia',
    'Thyroid',

    # Alcohol
    'Alcohol ever',

    # Smoking
    'Smoking habbit',

    # Central Obesity
    'Central Obesity'

]

for col in categorical_cols:

    if col in X.columns:

        le = LabelEncoder()

        X[col] = le.fit_transform(X[col])

        label_encoders[col] = le
```

```python
target_encoder = LabelEncoder()

y = target_encoder.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.3,

    random_state=42,

    stratify=y

)
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(

    X_train,
    y_train

)
```

```python
from sklearn.linear_model import LogisticRegression

# Create Model
model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

# Train Model
model.fit(X_train_smote, y_train_smote)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy : {accuracy*100:.2f}%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=target_encoder.classes_
))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=target_encoder.classes_,
    yticklabels=target_encoder.classes_
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Logistic Regression")

plt.show()
```

```python
y_pred_smote = model.predict(X_train_smote)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Convert SMOTE data to DataFrame (if needed)
X_train_smote_df = pd.DataFrame(X_train_smote, columns=X.columns)

# Calculate feature importance
importance = pd.DataFrame({
    'Feature': X_train_smote_df.columns,
    'Importance': np.mean(np.abs(model.coef_), axis=0)
})

# Sort for plotting
importance = importance.sort_values(by='Importance', ascending=True)

# Plot
plt.figure(figsize=(10,8))

plt.barh(
    importance['Feature'],
    importance['Importance'],
    color='steelblue',
    edgecolor='black'
)

plt.xlabel('Mean Absolute Coefficient', fontsize=12)
plt.ylabel('Features', fontsize=12)
plt.title('Feature Importance - Logistic Regression (SMOTE)', fontsize=14)

plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
